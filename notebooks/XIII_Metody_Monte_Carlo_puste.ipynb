Łatwo to nam przyjdzie jeżeli mamy możliwość generowania liczb pseudolosowych. Koncepcja stojąca za takimi generatorami jest zaskakująco prosta. Zauważmy, że kiedy spisujemy wyniki eksperymentu to tworzymy pewien ciąg (w naszych ostatnich eksperymentach głównie były to ciągi liczbowe). Ciągi te, jeżeli były odpowiednio długie, zdawały odpowiadające im testy statystyczne prawie na pewno. Jednak ciąg liczbowy w swojej naturze nie posiada informacji dotyczących źródła jego pochodzenia, czy został wytworzony z pewnej formuły, czy też spisując wzrost ludzi mieszkających w Krakowie. Tym samym nasuwa się pytanie, czy istnieje formuła, dla której otrzymamy ciąg liczb który również będzie spełniał testy statystyczne dla wybranego rozkładu. Odpowiedź oczywiście jest twierdząca, inaczej nie byłoby powodu by rozpisywać się na ten temat. Ale po co mielibyśmy takich generatorów szukać? Odpowiedzi jest wiele, na dwie jednak zamierzam zwrócić uwagę: wygoda, czyli fakt że możemy dzięki takiemu generatorowi utworzyć wiele próbek z danego rozkładu, bez konieczności tworzenia jakiegoś fizycznego eksperymentu, którego wynik będzie miał określony rozkład, zdecydowanie oszczędzając na czasie; oraz, paradoksalnie, determinizm, bo chociaż ciąg ma własności obiektu losowego, to powstaje bezpośrednio z formuły, zatem zaczynając z tych samych warunków początkowych otrzymamy dokładnie ten sam ciąg. Te dwie własności sprawiają, że takie generatory są wyjątkowo przydatne we wszelkich metodach Monte Carlo, gdzie często potrzebujemy wielu próbek z rozkładu danego typu, a jednocześnie chcemy możliwość odtworzenia tych próbek, żeby łatwiej wykrywać potencjalne błędy w naszych metodach.
