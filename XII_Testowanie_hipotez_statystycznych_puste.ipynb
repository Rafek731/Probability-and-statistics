Zajęcia dzisiejsze poświęcone są testowaniu hipotez statystycznych. Jest to metoda tak zwanego wnioskowania indukcyjnego, czyli próba wyciągnięcia pewnych wniosków na temat otaczającego nas świata bez bezpośredniego dowodu na zachodzącą zależność. Chociaż taka metoda wnioskowania może wydawać się naciągana jest z powodzeniem stosowana od stuleci w sprawach wszelkich. Dobrym przykładem takiego wnioskowania są sprawy sądowe. Oczywiście, jeżeli udało się przestępcę dosłownie złapać za rękę to takie wnioskowanie jest niepotrzebne, jednak w istotnej cześci przypadków takiej wygody nie mamy. Wówczas przedstawiane są argumenty, które mają nas przekonać do winy oskarżonego i jeżeli szanse na to, że nie jest on winowajcą są wystarczająco małe (w idealnym świecie) to uznaje się go za winnego zbrodni. Procedura sprawdzania hipotezy statystycznej opiera się na takim właśnie modelu i przyjęciu w nim roli oskarżającego.

Dostajemy pewną próbkę (oskarżonego) i pewne założenia na temat świata, które chcielibyśmy obalić (tak zwanne domniemanie niewinności). Te założenia pozwalają ustalić pewną statystykę której rozkład będzie nam znany (rozkład statystyki, a nie populacji z której próbka została wzięta). Wartość tej statystyki możemy wyliczyć ponieważ mamy próbkę. Co więcej możemy na podstawie naszych założeń policzyć z jakim prawdopodobieństwem ta statystyka osiągnie taką wartość (ten fragment to przekonywanie do winy). Jeżeli szanse na osiągnięcie tej wartości są odpowiednio małe to uznajemy nasze początkowe założenia za bezpodstawne (uznanie za winnego). Użyjmy jednak trochę formalniejszych nazw.

Podstawowe założenie dla danego problemu nazywamy hipotezą zerową. Będzie ona w zależności od problemu prowadziła do statystyki o pewnym znanym rozkładzie. Hipoteza zerowa to z reguły jest pewne bezpieczniejsze założenie, że może to co chcemy pokazać nie jest prawdą (domniemanie niewinności). Następnie do problemu dobieramy tak zwaną hipotezę alternatywną, czyli to co chcemy pokazać. Z reguły mamy trzy ich typy lewostronną, prawostronną  i obustronną. Pierwsza z nich mówi że hipoteza zerowa jest błędna i sprawdzana wartość jest jednak mniejsza, prawostronna twierdzi, że jest większa, zaś obustronna nie potrafi się zdecydować i mówi że po prostu jest różna. Podobnie jak w przypadku estymatorów decydujemy również z jakim prawdopodobieństwem chcielibyśmy się pomylić (tu nazywamy to poziomem istotności testu i oznaczamy przez $\alpha$). Na podstawie tego poziomu istotności oraz rozkładu danego przez nasze założenia wyliczamy tak zwaną wartość krytyczną, to jest wartość po której przekroczeniu prawdopodobieństwo otrzymania danej wartości jest zbyt małe i możemy w miarę bezpiecznie odrzucić hipotezę zerową na rzecz alternatywnej.
